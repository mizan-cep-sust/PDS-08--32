```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Given Polystyrene (PS) data
M_PS_table = np.array([1900, 867, 670, 411, 160, 98.2, 51, 19.8, 10.3, 3.7]) * 1e3 # in g/mol
Ve_one_fifth = np.array([23.75, 24.55, 25.20, 25.80, 27.30, 28.20, 29.40, 31.30, 32.50, 34.00])

# Calculate elution volume Ve = (Ve^(1/5))^5
Ve = Ve_one_fifth ** 5 # in cm^3

# Mark-Houwink parameters
K_PS = 4.9e-3  # cm^3/g
a_PS = 0.79

K_PDMS = 5.4e-3 # cm^3/g
a_PDMS = 0.77

# Universal Calibration:
# K_PS * M_PS^(a_PS + 1) = K_PDMS * M_PDMS^(a_PDMS + 1)
# We solve for M_PDMS:
M_PDMS = ((K_PS * (M_PS_table ** (a_PS + 1))) / K_PDMS) ** (1 / (a_PDMS + 1))

# Create a dataframe to view the calculated values clearly
df = pd.DataFrame({
    'Ve^(1/5) (cm^3/5)': Ve_one_fifth,
    'Ve (cm^3)': Ve,
    'M_PS (g/mol)': M_PS_table,
    'M_PDMS (g/mol)': M_PDMS,
    'log10(M_PDMS)': np.log10(M_PDMS)
})

print(df.to_string(index=False))

# Plot the calibration curve: log10(M_PDMS) vs. Ve
plt.figure(figsize=(10, 6))
plt.plot(Ve, np.log10(M_PDMS), 'o-', color='darkblue', label='PDMS Calibration')
plt.plot(Ve, np.log10(M_PS_table), 's--', color='gray', alpha=0.7, label='PS Calibration')
plt.title('GPC Calibration Curve in Chloroform at 35°C')
plt.xlabel('Elution Volume, $V_e$ ($cm^3$)')
plt.ylabel('$\log_{10}(M)$')
plt.grid(True, which="both", ls="--")
plt.legend()
plt.show()
```
